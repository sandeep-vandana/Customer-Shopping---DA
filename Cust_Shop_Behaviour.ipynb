```python
import pandas as pd
import numpy as np
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head(2)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.groupby('Category')['Review Rating'].mean()
```

```python
df['Review Rating'] = df['Review Rating'].fillna(df.groupby('Category')['Review Rating'].transform('mean'))
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns
```

```python
df.columns = df.columns.str.replace(' ','_')
```

```python
df.head()
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount_'})
```

```python
df.columns
```

```python
df['age']
```

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
```

```python
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
```

```python
df['age_group'].value_counts()
```

```python
df[['age','age_group']].head(4)
```

```python
df['frequency_of_purchases'].value_counts()
```

```python
# create new column purchase_frequency_days
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
```

```python
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(5)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
df.to_csv('Cust_Shop_Behaviour.csv')
```
